### 8.2 自訂函式 (User-defined Functions)

下面範例程式可以幫助您快速了解系統當前的記憶體使用狀況，包括總記憶體、可用記憶體、已用記憶體以及記憶體使用率。

說明：

1. psutil 模組
- 用於獲取系統資源使用情況
- psutil.virtual_memory() 返回系統記憶體使用情況的命名元組
2. convert_bytes 函數
- 功能：將位元組(bytes)轉換為GB
- 參數：size - 要轉換的位元組數
- 返回值：轉換後的GB數值（浮點數）
3. 記憶體變數說明：
- memory.total：系統總記憶體（位元組）
- memory.available：可用記憶體（位元組）
- memory.used：已使用記憶體（位元組）
- memory.percent：記憶體使用百分比
4. 輸出格式：
- 記憶體大小顯示到小數點後三位
- 使用f-string進行格式化輸出


````python
import psutil

# 獲取系統記憶體資訊
memory = psutil.virtual_memory()

def convert_bytes(size):
    """將位元組(bytes)轉換為GB
    
    Args:
        size (int): 位元組大小
        
    Returns:
        float: 轉換後的GB數值
    """
    # 將位元組轉換為GB (1024^3 bytes = 1 GB)
    gb = size / (1024 ** 3)
    return gb

# 轉換記憶體單位
total_gb = convert_bytes(memory.total)       # 系統總記憶體
available_gb = convert_bytes(memory.available)  # 可用記憶體
used_gb = convert_bytes(memory.used)         # 已使用記憶體

# 輸出記憶體使用情況
print(f"總記憶體: {total_gb:.3f} GB")
print(f"可用記憶體: {available_gb:.3f} GB")
print(f"已用記憶體: {used_gb:.3f} GB")
print(f"記憶體使用率: {memory.percent}%")
````

### 8.3 Positioanl and Keyword Arguments

Python functions can accept arguments in two ways: positional arguments and keyword arguments. Positional arguments are arguments that are passed to a function in the order in which they are defined. Keyword arguments are arguments that are passed to a function by name. 

注意事項:

- 位置參數必須按照函數定義的順序傳遞
- 必要的位置參數必須在呼叫時提供
- 有預設值的參數應該放在沒有預設值的參數之後
- 位置參數在函數調用時不能重複指定

範例展示:

```python
def greet(name, greeting):
    """這是一個簡單的問候函數
    
    Args:
        name (str): 要問候的人名
        greeting (str): 問候語
        
    Returns:
        str: 完整的問候訊息
    """
    return f"{greeting}, {name}!"

# 使用位置參數呼叫函數
message = greet("小明", "你好")  # 參數按照定義的順序傳遞
print(message)  # 輸出: 你好, 小明!
```


````python
# 多個位置參數
def calculate_rectangle_area(length, width):
    """計算長方形的面積
    
    Args:
        length (float): 長度
        width (float): 寬度
        
    Returns:
        float: 長方形面積
    """
    return length * width

# 呼叫函數時，參數必須按照定義的順序
area = calculate_rectangle_area(5.5, 3.2)
print(f"長方形面積: {area} 平方單位")  # 輸出: 長方形面積: 17.6 平方單位

# 位置參數的強制性
# 位置參數是必須的，如果呼叫時沒有提供足夠的參數，會引發 TypeError：
try:
    area = calculate_rectangle_area(5.5)  # 引發 TypeError
except TypeError as e:
    print(e)

# 位置參數與預設值結合
# 位置參數也可以有預設值，但有預設值的參數必須放在沒有預設值的參數之後：
def power(base, exponent=2):
    """計算冪次方
    
    Args:
        base (int or float): 底數
        exponent (int or float, optional): 指數，預設為2
        
    Returns:
        int or float: 計算結果
    """
    return base ** exponent

# 使用預設指數
print(f'power(3)\t={power(3)}')      # 輸出: 9 (3的2次方)

# 指定指數
print(f'power(3, 3)\t={power(3, 3)}')   # 輸出: 27 (3的3次方)

# positional arguments becomes keyword arguments
print(f'power(base=3, exponent=3)\t={power(base=3, exponent=3)}')

# The order of keyword arguments becomes irrelevant
print(f'power(exponent=3, base=3)\t={power(exponent=3, base=3)}')
````

參數說明：

1. 僅限位置參數 (Positional-only parameters)
- 在參數清單中使用 / 標記
- 必須在函數呼叫時提供
- 不能使用關鍵字參數的形式傳入

2. 可選的位置/關鍵字參數 (Positional or keyword parameters)
- 位於 / 和 * 之間
- 可以有預設值
- 可以使用位置或關鍵字參數形式傳入

3. 僅限關鍵字參數 (Keyword-only parameters)
- 在參數清單中使用 * 標記
- 必須使用關鍵字參數形式傳入
- 提高程式碼可讀性，強制呼叫者明確指定參數名稱

這種參數設計方式在 Python 3.8+ 中特別有用，可以建立更清晰、更安全的 API。

````python
def process_data(name, age,/, sex='男', *, city):
    """處理使用者資料的函數
    
    這個函數展示三種參數類型：
    - 僅限位置參數 (name)
    - 可選的位置/關鍵字參數 (age)
    - 僅限關鍵字參數 (city)
    
    Args:
        name (str): 使用者姓名 (僅限位置參數)
        age (int, optional): 使用者年齡，預設為25 (位置參數)
        sex (str, optional): 使用者性別，預設為男 (可選的位置/關鍵字參數)
        city (str): 使用者所在城市 (僅限關鍵字參數)
        
    Returns:
        dict: 包含處理後的使用者資料
    """
    return {
        "name": name,
        "age": age,
        "sex": sex,
        "city": city,
        "is_adult": age >= 18
    }

# 正確的呼叫方式
# 1. 只提供必要參數
try:
    # 會出錯，因為 age 是必須的位置參數
    print(f'process_data("王小明", city="台北") = \n\t錯誤: 缺少位置參數') 
    result = process_data("王小明", city="台北")
except Exception as e:
    print(e) 

# 2. 使用位置參數
# 預設的 'sex'
result1 = process_data("王小明", 30, city="台北")
print(f'process_data("王小明", 30, city="台北") = \n\t{result1}') 

# 3. 使用關鍵字參數 (age)
try:
    # 會出錯，因為 'age' 是必須的位置參數
    print(f'process_data("王小明", age=30, city="台北") = \n\t錯誤: age 非位置參數') 
    result2 = process_data("王小明", age=30, city="台北")
except Exception as e:
    print(e) 

# 4. 錯誤: 缺少關鍵字參數 'city'
try:
    print(f'process_data("王小明", 30, "台北") = \n\t錯誤: 缺少關鍵字參數') 
    result3 = process_data("王小明", 30, "台北")  
except Exception as e:
    print(e) 
````

### 9.3 行星拉格朗日點（Lagrange Points） 

拉格朗日點是兩個大天體（例如太陽與地球、地球與月球）在互相引力加上離心力的作用下，形成的一種特殊平衡位置。

把一個小物體（太空船、小行星、太空塵埃等）放在這些點上，它就能相對於這兩個大天體保持相對靜止，跟著它們一起繞圈，幾乎不需要額外推進力。

在太陽－行星系統中，總共有五個拉格朗日點，標記為 L1、L2、L3、L4、L5。

下面範例程式碼繪製日地系統拉格朗日點，這個視覺化清楚地展示了日地系統中的五個拉格朗日點位置，有助於理解這些特殊點在太空中的相對位置。

1. 基本設定：
- 設定太陽在原點 (0,0)，地球在 (1,0) AU 處
- 使用 AU (天文單位) 作為距離單位，1 AU = 地球到太陽的平均距離
2. 拉格朗日點計算：
- L1：位於地球朝向太陽的一側，約 0.99 AU
- L2：位於地球背向太陽的一側，約 1.01 AU
- L3：位於太陽另一側，約 -1 AU
- L4 和 L5：與太陽和地球形成等邊三角形
3. 視覺化：
- 藍色圓點：太陽
- 黃色圓點：地球
- 紅色 X：拉格朗日點
- 虛線圓：地球軌道
4. 圖表元素：
- 網格線：幫助讀取座標
- 圖例：說明各元素代表意義
- 座標軸標籤：標示距離單位
- 距離標記：X 軸上的 AU 標記

````python
import matplotlib.pyplot as plt
import numpy as np

# 設定中文字體
plt.rcParams['font.sans-serif'] = ['Arial Unicode MS', 'Microsoft JhengHei', 'SimHei', 'sans-serif']
plt.rcParams['axes.unicode_minus'] = False  # 正確顯示負號
#
# 太陽系座標設定
# 假設太陽在原點 (0,0)，地球在 (1,0)，距離單位為 AU (天文單位)
sun = (0, 0)      # 太陽座標
earth = (1, 0)    # 地球座標

# 拉格朗日點近似位置計算
# L1 ≈ 0.99, L2 ≈ 1.01, L3 ≈ -1 (相對於太陽的距離，單位AU)
# L4 和 L5 形成等邊三角形，與太陽和地球夾角60度
mu = 3e-6          # 地球與太陽的質量比，約為 3×10⁻⁶

# 計算拉格朗日點位置
L1_approx = 1 - (mu/3)**(1/3)   # L1 點位置，約 0.990 AU
L2_approx = 1 + (mu/3)**(1/3)   # L2 點位置，約 1.010 AU
L3_approx = -1                   # L3 點位置，在太陽另一側

# 計算 L4 和 L5 點座標（等邊三角形頂點）
theta = np.deg2rad(60)  # 60度轉換為弧度
L4_x = 0.5              # L4 點的 x 座標
L4_y = np.sin(theta)    # L4 點的 y 座標
L5_x = 0.5              # L5 點的 x 座標
L5_y = -L4_y            # L5 點與 L4 點對稱於 x 軸

# 創建圖形
fig, ax = plt.subplots(figsize=(9, 7))

# 繪製太陽和地球
ax.plot(sun[0], sun[1], 'bo', markersize=20, label='太陽 (Sun)')
ax.plot(earth[0], earth[1], 'yo', markersize=12, label='地球 (Earth)')

# 繪製地球軌道（圓形近似）
theta_orbit = np.linspace(0, 2*np.pi, 200)
ax.plot(np.cos(theta_orbit), np.sin(theta_orbit), 'b--', lw=0.8, alpha=0.5, 
        label='地球軌道 (Earth\'s Orbit)')

# 定義五個拉格朗日點的座標和標籤
lagrange_points = [
    (L1_approx, 0, 'L1'),      # 第一拉格朗日點
    (L2_approx, 0, 'L2'),      # 第二拉格朗日點
    (L3_approx, 0, 'L3'),      # 第三拉格朗日點
    (L4_x, L4_y, 'L4'),        # 第四拉格朗日點
    (L5_x, L5_y, 'L5')         # 第五拉格朗日點
]

# 繪製拉格朗日點
for x, y, label in lagrange_points:
    ax.plot(x, y, 'x', markersize=8, color='red')
    # 調整標籤位置
    if label == 'L1':
        ax.text(x-0.1, y+0.08, label, fontsize=12, color='blue', 
                bbox=dict(facecolor='white', alpha=0.7, edgecolor='none', pad=1))
    else:
        ax.text(x+0.02, y+0.08, label, fontsize=12, color='red',
                bbox=dict(facecolor='white', alpha=0.7, edgecolor='none', pad=1))

# 添加圖表標題和標籤
ax.set_title('日地系統拉格朗日點 (Sun-Earth Lagrange Points)', fontsize=14, pad=20)
ax.set_xlabel('X 座標 (AU)', fontsize=12)
ax.set_ylabel('Y 座標 (AU)', fontsize=12)

# 設置座標軸範圍和比例
ax.set_xlim(-1.5, 1.5)
ax.set_ylim(-1.2, 1.2)
ax.set_aspect('equal')  # 確保 x 和 y 軸比例相同

# 添加網格和圖例
ax.grid(True, alpha=0.3)
ax.legend(loc='upper right')

# 添加參考線
ax.axhline(0, color='k', linestyle='-', alpha=0.1)
ax.axvline(0, color='k', linestyle='-', alpha=0.1)

# 添加距離標記
for x in np.arange(-1.5, 2.0, 0.5):
    ax.axvline(x, color='gray', linestyle=':', alpha=0.2)
    if x != 0:  # 跳過原點
        ax.text(x, -1.3, f'{x:.1f} AU', ha='center', va='top', fontsize=8)

# 顯示圖形
plt.tight_layout()
plt.show()
````
