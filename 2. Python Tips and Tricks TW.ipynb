清晰的名稱讓您的代碼不言自明，減少註解的需求。

### 1.3 Python 的 match 語句說明
match 是 Python 3.10 引入的結構化模式匹配語法，它提供了一種更強大、更直觀的方式來進行條件分支處理，特別適合處理複雜的數據結構。

基本語法:
```
match subject:
    case pattern1:
        # 處理 pattern1 的情況
    case pattern2:
        # 處理 pattern2 的情況
    case _:
        # 預設情況（類似於 switch 中的 default）
```
優點:

- 比傳統的 if-elif-else 鏈更簡潔
- 支持結構化數據的深度匹配
- 更易讀，特別是在處理複雜的數據結構時
- 可以與類型提示結合使用，提高代碼的可讀性和可維護性

注意事項:

- match 是 Python 3.10 的新特性，舊版本不支援
- 模式匹配是按順序進行的，第一個匹配的模式會被執行
- "_" 是萬用字元，可以匹配任何值
- 模式可以嵌套，允許複雜的數據結構匹配
- match 語句特別適合處理 JSON 數據、解析命令列參數、處理 AST（抽象語法樹）等場景。

程式碼說明：

1. 函數功能：
- describe_type() 函數使用 Python 3.10+ 的 match-case 語法來判斷輸入值的類型
- 根據不同類型返回對應的中文描述
2. 支援的類型：
- 布林值（bool）：顯示「真」或「假」
- 整數（int）：區分正整數和負整數
- 字串（str）：顯示長度和內容
- 列表（list）：顯示元素數量
- 集合（set）：顯示不重複元素的數量
- 元組（tuple）：顯示元素數量
- 字典（dict）：顯示鍵值對數量
- 空值（None）
- 其他類型：顯示類型名稱
3. 測試案例：
- 包含各種常見的 Python 數據類型
- 每個測試案例都展示了不同的類型判斷情況
- 輸出結果會顯示原始值及其對應的類型描述
4. 輸出格式：
- 每行顯示一個測試值及其類型描述
- 使用字串格式化對齊輸出，提高可讀性

````python
def describe_type(value):
    """
    根據輸入值的類型返回相應的描述字串
    
    Args:
        value: 要檢查類型的值，可以是任何Python對象
    
    Returns:
        str: 描述值類型和特性的字串
    """
    match value:
        case bool():
            return f"布林值: {'真' if value else '假'}"
        case int() if value >= 0:
            return f"非負整數: {value}"
        case int():  # 負整數
            return f"負整數: {value}"
        case str():
            length = len(value)
            if length == 0:
                return "空字串"
            return f"長度為 {length} 的字串: '{value}'"
        case list():
            length = len(value)
            return f"包含 {length} 個元素的列表"
        case set():
            length = len(value)
            return f"包含 {length} 個不重複元素的集合"
        case tuple():
            length = len(value)
            return f"包含 {length} 個元素的元組"
        case dict():
            length = len(value)
            return f"包含 {length} 個鍵值對的字典"
        case None:
            return "空值 (None)"
        case _:
            return f"其他類型: {type(value).__name__}"

# 測試案例
test_values = [
    42,                     # 正整數
    -7,                     # 負整數
    True,                   # 布林值 True
    False,                  # 布林值 False
    "",                     # 空字串
    "hello",                # 一般字串
    [1, 2, 3],              # 列表
    {1, 2, 3, 3},           # 集合（自動去重）
    (10, 20),               # 元組
    {"name": "Alice", "age": 25},  # 字典
    None,                   # 空值
    3.14                    # 浮點數（會進入預設情況）
]

print("類型判斷範例：\n" + "="*30)
for val in test_values:
    print(f"{val!r:30} → {describe_type(val)}")
````

### 1.4 使用 _（底線）作為忽略的變數或佔位符

當您有一個數值變量，其值不容易閱讀時，可以使用 _ 來表示千位分隔，以提高可讀性。

```python
# 不好的寫法
num = 1000000

# 好的寫法
num = 1_000_000
```

或者當您不需要使用循環變量時，可以使用 _ 來表示這個變量是被故意忽略的。這是一個簡單但強大的約定。

```python
for _ in range(3):
    print("Hello, World!")
```

````python
num1 = 10000000000
num2 = 100000000

total = num1 + num2
print(f"總和 = {total}")
````

````python
# 使用 "_" 提升可讀性（無操作）
num1 = 10_000_000_000
num2 = 100_000_000

total = num1 + num2
print(f"總和 = {total}")

# 使用 "_" 表示未使用的循環計數器
for _ in range(3):
    print("Hello World!")
````

````python
# 使用 ',' 格式化提升報表可讀性
print(f'總和 = {total:,}')
````

### 1.9 字典處理

- 遍歷字典

````python
student = {
        'name': 'John',
        'age': 25,
        'courses': ['Math', 'ComSci']
    }
# 使用 items() 方法遍歷
print("使用 items() 方法遍歷:")
for key, value in student.items():
    print(f"\t{key}: {value}")
    
# 使用 keys() 方法遍歷
print("\n使用 keys() 方法遍歷:")
for key in student.keys():
    print(f"\t{key}")
    
# 使用 values() 方法遍歷
print("\n使用 values() 方法遍歷:")
for value in student.values():
    print(f"\t{value}")
````

### 2.1 使用 `Context Manager` 讀取檔案

常見的讀取方法有：

    1. f.read()
    2. f.readlines()
    3. f.readline()
    4. f.seek(位置)
    5. f.tell()
    
以及屬性：

    1. f.mode
    2. f.closed

````python
# 使用 'with' 上下文管理器讀取檔案
# 不僅適用於檔案，還適用於其他資源，如線程

with open('sample/test.txt', 'r') as f:
    
    print(f'檔案物件:\t {f.name}')
    print(f'---> 開啟模式:\t {f.mode}\n')

    # 將整個檔案內容讀取到一個字串變數中
    file_contents = f.read()
    print(f'已將整個檔案讀取到一個字串變數中:\n{file_contents}\n')


# 上下文管理器在離開 with 區塊後會自動關閉檔案
print(f'---> 上下文管理器已結束。檔案是否已關閉?\t {f.closed}\n')

try:
    # 嘗試在檔案關閉後讀取
    no_contents = f.read()

except ValueError: 
    print('f.read(): 在已關閉的檔案上進行 I/O 操作。\n')
except:
    print('發生其他錯誤。\n')

    
words = file_contents.split(' ')
print(f'查看 "test.txt" 檔案中的單字:\n{words}\n')

word_count = len(words)
print(f'檔案 "test.txt" 中共有 {word_count} 個單字。\n')
````

### 2.2 更多關於檔案讀取方法的範例

````python
# 更多讀取檔案的方法
with open('sample/test.txt', 'r') as f:
    
    # 將檔案內容讀取到一個行列表中
    f_contents = f.readlines()
    print(f'一次讀取所有行到列表中:\n{f_contents}\n')

    # 回到檔案開頭
    f.seek(0)

    # 一次讀取一行
    f_contents = f.readline()
    print(f'讀取第一行到列表中:\n{f_contents}')

    f_contents = f.readline()
    print(f'讀取第二行到列表中:\n{f_contents}')

    # 忽略換行符
    f_contents = f.readline()
    print(f'讀取第三行到列表中，忽略換行:\n{f_contents}', end='')
    f_contents = f.readline()
    print(f'讀取第四行到列表中，忽略換行:\n{f_contents}', end='')

    # 使用 for 迴圈逐行讀取，直到檔案結尾
    for line in f:
        print(f'使用 for 迴圈逐行讀取，直到檔案結尾...\n->{line}<-\n', end='')

    # 回到檔案開頭 (BOF)
    f.seek(0)

    # 使用 while 迴圈每次讀取指定字節數，直到檔案結尾 (EOF)
    size_to_read = 100
    f_contents = f.read(size_to_read)
    while len(f_contents) > 0:
        print(f'\n使用 while 迴圈，每次讀取 {size_to_read} 字節，直到檔案結尾...\n->{f_contents}<-\n', end='')
        f_contents = f.read(size_to_read)

    # 再次回到檔案開頭
    f.seek(0)

    # 以較小的區塊讀取，每次 10 個字元
    size_to_read = 10
    f_contents = f.read(size_to_read)
    print(f'讀取 {size_to_read} 字節，忽略換行:\n->{f_contents}<-\n', end='')
    print(f'當前檔案物件位置 = {f.tell()}\n')

    f.seek(0)
    f_contents = f.read(size_to_read)
    print(f'回到檔案開頭，讀取 {size_to_read} 字節，忽略換行:\n->{f_contents}<-\n', end='')

    print(f'當前檔案物件位置 = {f.tell()}\n')
````

````python
# 複製檔案到另一個檔案
with open("sample/test.txt", "r") as rf:
    with open("test_copy.txt", "w") as wf:
        for line in rf:
            wf.write(line)

file_read = "test_copy.txt"
with open(file_read, "r") as rf:
    print(f'開啟 {file_read} 檔案:\n')
    for line in rf:
        print(f'逐行讀取，直到檔案結尾...\n->{line}<-\n', end='')

# 複製圖片：以二進位模式讀取/寫入
with open("sample/Olisan.JPG", "rb") as rf:
    with open("sample/Olisan_copy.JPG", "wb") as wf:
        for line in rf:
            wf.write(line)

# 以區塊方式複製圖片
with open("sample/Olisan.JPG", "rb") as rf:
    with open("sample/Olisan_copy2.JPG", "wb") as wf:
        chunk_size = 4096
        rf_chunk = rf.read(chunk_size)
        
        while len(rf_chunk) > 0:
            wf.write(rf_chunk)
            rf_chunk = rf.read(chunk_size)

# 顯示所有圖片
from PIL import Image
img = Image.open('sample/Olisan.JPG')
img.show()
img_copy = Image.open('sample/Olisan_copy.JPG')
img_copy.show()
img_copy2 = Image.open('sample/Olisan_copy2.JPG')
img_copy2.show()
````

## 3. 使用 `try-except-else-finally` 進行錯誤處理

使用 `try/except` 優雅地處理運行時錯誤。
它可以防止您的代碼意外崩潰。

````python
try:
    # 可能會引發異常的代碼
    result = 10 / 0
except ZeroDivisionError:
    # 處理特定的異常
    print("錯誤：不能除以零")
else:
    # 如果沒有發生異常，則運行此代碼
    print("else: 除法運算成功")
finally:
    # 無論是否發生異常，此代碼都會執行
    print("finally: 這行代碼總會執行")
````

## 4. 使用 `enumerate` 進行計數

在遍歷列表時，使用 enumerate() 而不是手動計數器。它會自動提供索引和值，提高代碼清晰度。

````python
# 使用計數器的舊方法
names = ['Corey', 'Michael', 'Travis', 'Dave']

index = 0
for name in names:
    print(index, name)
    index += 1
````

````python
# 使用 enumerate 的更簡潔代碼
names = ['Corey', 'Michael', 'Travis', 'Dave']

for index, name in enumerate(names):
    print(index, name)
````

````python
# 另一個使用 enumerate 的更簡潔代碼，從 1 開始計數
names = ['Corey', 'Michael', 'Travis', 'Dave']

# 將起始索引設為 1
for index, name in enumerate(names, start=1):
    print(index, name)
````

## 5. 使用 `zip` 處理多個物件

使用 zip() 整齊地組合多個列表。非常適合並行遍歷相關數據。

````python
names = ['Corey', 'Michael', 'Travis', 'Dave']
heroes = ['Spiderman', 'Superman', 'Deadpool', 'Batman']
for index, name in enumerate(names):
    hero = heroes[index]
    print(f'{name} 實際上是 {hero}')
````

````python
names = ['Corey', 'Michael', 'Travis', 'Dave']
heroes = ['Spiderman', 'Superman', 'Deadpool', 'Batman']

# 使用 zip 的更簡潔代碼
for name, hero in zip(names, heroes):
    print(f'{name} 實際上是 {hero}')
````

````python
names = ['Corey', 'Michael', 'Travis', 'Dave']
heroes = ['Spiderman', 'Superman', 'Deadpool', 'Batman']
universes = ['Marvel', 'DC', 'Tokyo', 'Taiwan']

# 使用 zip 合併多個列表的更簡潔代碼
for name, hero, universe in zip(names, heroes, universes):
    print(f'{name} 實際上是來自 {universe} 的 {hero}')
````

## 6. 使用解包(unpacking)將物件賦值給變數

它允許您在一行程式碼中將值分配給多個變數。

````python
# 元組的基本解包
a, b = (1, 2)
print(f"a: {a}")
print(f"b: {b}")
````

````python
# 解包列表
x, y = [10, 20]
print(f"x: {x}, y: {y}")
````

````python
# 在函數參數中進行解包
def add(a, b, c):
    return a + b + c

args = (1, 2, 3)
print(f"add(*args) = {add(*args)}")
````

````python
# 解包字串（任何可迭代對象 iterable objects，字串由一系列字符組成）
a, b, c = "abc"
print(f"a={a}, b={b}, c={c}")
````

````python
# 解包字典，使用 ** 來解包鍵值對
def greet(name, age):
    print(f"你好，{name}。你今年 {age} 歲。")

person = {"name": "John", "age": 30}
greet(**person)

# 等同於
greet(name="John", age=30)
````

````python
# 從字典中解包鍵或值
d = {'a': 1, 'b': 2, 'c': 3}

k1, k2, k3 = d.keys()
v1, v2, v3 = d.values()

print(f"k1={k1}, k2={k2}, k3={k3}")
print(f"v1={v1}, v2={v2}, v3={v3}")
````

````python
# 使用 * 進行擴展解包 (Python3+)
a, b, *c, d = (1, 2, 3, 4, 5, 6)
print(f"a: {a}")
print(f"b: {b}")
print(f"c: {c}")
print(f"d: {d}")
````

````python
# 使用 *_ 來忽略多餘的解包元素 (Python3+)
a, b, *_, d = (1, 2, 3, 4, 5, 6)
print(f"a: {a}")
print(f"b: {b}")
print(f"d: {d}")
````

````python
# 巢狀解包
coords =[(1,2),(3,4),(5,6)]
x,y = zip(*coords)
print(f"x: {x}")
print(f"y: {y}")
````

````python
# 更複雜的巢狀列表
data = ['Michael', (25, 'Engineer')]

# 解包
name, (age, profession) = data
print(f"姓名: {name}, 年齡: {age}, 職業: {profession}")
````

## 7. 使用 `setattr`/`getattr` 動態設定/獲取物件屬性

動態設定和獲取物件的屬性。

````python
class Person():
    pass

person = Person()

# 設定屬性和值
person.first = "Michael"
person.last = "Kao"

print(f'first: {person.first}, last: {person.last}')
````

````python
class Person():
    pass

person = Person()

# 使用變數值動態設定屬性
key = "first"
value = "Michael"  
person.last = "Kao"

setattr(person, key, value)

print(f'first: {person.first}, last: {person.last}')
````

````python
class Person():
    pass

person = Person()

# 使用變數值動態設定屬性
key = "first"
value = "Michael"  
person.last = "Kao"

setattr(person, key, value)

# 使用變數值獲取屬性值
first = getattr(person, key)
print(f"first: {first}")
````

````python
class Person():
    pass

person = Person()

person_dict = {'first':'Michael', 'last':'Kao'}

for key, value in person_dict.items():
    setattr(person, key, value)

print(f'{person.first} {person.last}')

for key in person_dict.keys():
    print(getattr(person, key))
````

## 11. 使用 `Iterables` 進行迭代

可迭代物件(Iterable)是一個可以進行迭代的物件。當傳遞給 iter() 方法時，它會生成一個迭代器(Iterator)。

迭代器是一個物件，用於通過 __next__() 方法對可迭代物件進行迭代。

迭代器具有 __next__() 方法，該方法返回物件的下一個項目。請注意，每個迭代器都是一個可迭代物件，但並非每個可迭代物件都是迭代器。

例如，列表(list)是可迭代的，但列表本身不是迭代器。可以通過 iter() 函數從可迭代物件創建迭代器。為了實現這一點，物件的類需要一個返回迭代器的 __iter__ 方法，或者一個從 0 開始的連續索引的 __getitem__ 方法。

````python
for element in ['c123', 142_857, 'Michael', 'Kao']:
    print(element, end=" ")
print()

for element in ('c123', 142_857, 'Michael', 'Kao'):
    print(element, end=" ")
print()

for letter in 'Michael Kao':
    print(letter, end=" ")
print()

for byte in b'Binary':
    print(byte, end=" ")
print()

# 上面展示的所有物件都是可迭代的
# 然而，整數型別是不可迭代的，如下所示
c = 142_857

# 必須轉換為字串才能讓整數可迭代
for digit in str(c):
    print(digit, end=" ")
print()

# 或者轉換為列表讓整數可迭代
digits = [int(d) for d in str(c)]
for digit in digits:
    print(digit, end=" ")
print()

# 導入必要的模組
import itertools

# 建立牌面數字清單：2-10 加上 J, Q, K, A
ranks = list(range(2, 11)) + ["J", "Q", "K", "A"]
# 將所有元素轉換為字串
ranks = [str(rank) for rank in ranks]

print(f'牌面數字: {ranks}')

# 定義四種花色
suits = ['♠', '♥', '♦', '♣']

# 使用 itertools.product 建立所有可能的牌面組合（花色 + 數字）
deck = [card for card in itertools.product(suits, ranks)]

# 遍歷並顯示前52張牌
for (index, card) in enumerate(deck):
    # 顯示牌組中的每一張牌，編號從1開始
    print(f'{1+index}: {card}')
    # 安全機制：最多只顯示52張牌
    if index >= 51:  # 因為索引從0開始，所以51表示第52張牌
        break

# 計算從52張牌中選取5張的所有可能組合
# 注意：這個計算量非常大（2,598,960種組合）
hands = [hand for hand in itertools.combinations(deck, 5)]
print(f'五張牌的撲克牌組合共有 {len(hands):,} 種。')
````
