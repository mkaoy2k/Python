## 4. Lists

- Adding elements to a list
- Removing elements from a list
- Accessing elements in a list
- Slicing a list
- Insert an element into a list
- Concatenating lists
- Nested lists
- Sorting a list
- Copying a list


````python
countries = ['United States', 'India', 'Taiwan', 'Brazil']
print(countries)

# add an element to the list
countries.append('Japan')
print(countries)

# remove an element from the list
countries.remove('Brazil')
print(countries)
````

````python
# More ways to remove items from a list

countries = ['United States', 'India', 'Taiwan', 'Brazil']
print(countries)

# Remove the last item
removed_country = countries.pop()
print(removed_country)

countries.pop(-1)
print(countries)

del countries[0]
print(countries)
````

````python
# access an element in a list (zero indexed)
countries = ['United States', 'India', 'Taiwan', 'Brazil']
print(countries[0])
````

````python
# reverse indexed
countries = ['United States', 'India', 'Taiwan', 'Brazil']
print(countries[-1])
````

````python
countries = ['United States', 'India', 'Taiwan', 'Brazil']
print(countries[-4])
````

````python
# slicing a list[start:stop]
countries = ['United States', 'India', 'Taiwan', 'Brazil']
print(countries[0:2])
````

````python
countries = ['United States', 'India', 'Taiwan', 'Brazil']
print(countries[1:])
````

````python
countries = ['United States', 'India', 'Taiwan', 'Brazil']
print(countries[:2])
````

````python
# insert an element in the list
countries = ['United States', 'India', 'Taiwan', 'Brazil']
countries.insert(0, 'Spain')
print(countries)
````

````python
# Concatenation of lists
countries = ['United States', 'India', 'Taiwan', 'Brazil']
countries_2 = ["Finland", "Sweden", "Denmark", "Norway", "Iceland"]
print(countries + countries_2)
````

````python
# nested list of lists
countries = ['United States', 'India', 'Taiwan', 'Brazil']
countries_2 = ['China', 'Japan', 'South Korea', 'Vietnam']
nested_list = [countries, countries_2]
print(nested_list)
````

````python
# sorting a list
numbers = [4, 3, 10, 9, 8, 2]
print(numbers)
numbers.sort()
print(numbers)
````

````python
# sorting in reverse order
numbers = [4, 3, 10, 9, 8, 2]
numbers.sort(reverse=True)
print(numbers)
````

````python
# update an element in the list
numbers = [4, 3, 10, 9, 8, 2]
print(numbers)

numbers[0] = 1000
print(numbers)
````

````python
# copy a list
countries = ['United States', 'India', 'Taiwan', 'Brazil']
new_list = countries[:]
print(new_list)
````

````python
# another way to copy a list
new_list2 = countries.copy()
print(new_list2)
````

### 5.1 Copy a dictionary

````python
my_data = {'name': 'Michael', 'age': 63, 'height': 165}
print(my_data)

new_dict = my_data.copy() # copy by data
print(new_dict)

new_dict2 = my_data # copy by reference
new_dict2
````

````python
my_data.update({'height':180})
print(my_data)
print(new_dict) # does not affect the other copy of data
new_dict2 # reference to the same copy of data
````

### 5.2 Remove an element from dictionary

````python
my_data = {'name': 'Michael', 'age': 63, 'height': 180}
print(my_data)
my_data.pop('name') # pop by key
my_data
````

````python
del my_data['age'] # delete by key
my_data
````

````python
my_data.clear() # remove all elements
my_data
````

## 6. If Statement

Syntax:
```python
if (condition):
    code block
else:
    code block
```

````python
age = 10

if age >= 18:
    print("You're an adult!")
elif age >= 13:
    print("You're a teenager.")
else:
    print("You're a kid.")
````

### 8.2 User-defined Functions

````python
import psutil

memory = psutil.virtual_memory()

def convert_bytes(size):
    # Convert bytes to GB
    gb = size / (1024 ** 3)
    return gb

total_gb = convert_bytes (memory.total)
available_gb = convert_bytes(memory.available)
used_gb = convert_bytes (memory.used)

print(f"Total RAM: {total_gb:.3f} GB ")
print(f"Available RAM: {available_gb:.3f} GB ")
print(f"Used RAM: {used_gb:.3f} GB ")
print(f"RAM Usage: {memory.percent}%")
````
