## 1. How to make Python code more readable

### 1.1 Use Descriptive Variable and Fuction Names

Choose variable and function names that describe what they store or do.
For example, instead of a = 10, write user_age = 10.
Clear names make your code self-explanatory and reduce the need for comments.


````python
# Bad
a = 5
b = 2

# Good
total = 5
sub_total = 2
print("Total:   ", total)
print("Subtotal:", sub_total)
````

````python
# Bad
def a():
    return 5

def b():
    return 2

# Good
def calculate_total():
    return 5

def calculate_sub_total():
    return 2

print("Calling calculate_total()     =", calculate_total())
print("Calling calculate_sub_total() =", calculate_sub_total())
````

## 2. Use Context Manager to Manage Resources

When we read from or write to a file, we sometimes forget to close it at the end. It certainly waste system resources, such as file object and buffer, once allocated and not being able to reuse.

Python has a context manager via "with" statement to manage the resouces for you once you exit the execution of "with"-bolck. Always use with open() for file operations. It ensures files are closed automatically, preventing leaks or errors.


Here is the sample file, called "test.txt", containing the following:

    1) This is a test file
    2) With multiple lines of data...
    3) Third line
    4) Fourth line
    5) Fifth line
    6) Sixth line
    7) Seventh line
    8) Eighth line
    9) Ninth line
    10) Tenth line


````python
# Open File Object Modes:
# f = open("test.txt", "r")  # for read
# f = open("test.txt", "w") # for write
# f = open("test.txt", "a") # for append
# f = open("test.txt", "r+") # for read and write

f = open('test.txt', 'r')
print(f'File object:\t {f.name}')
print(f'---> Open Mode:\t {f.mode}\n')

file_contents = f.read()
f.close()
print(f'---> is closed?\t {f.closed}\n')


print(f'read the whole file into a string variable at a time:\n{file_contents}\n')

words = file_contents.split(' ')
print(f'words in the "test.txt" file:\n{words}\n')

word_count = len(words)
print(f'There are {word_count} words in the "test.txt" file.\n')
````

### 2.1 Read a File using Context Manager


The common reading methods are:

    1. f.read()
    2. f.readlines()
    3. f.readline()
    4. f.seek(position)
    5. f.tell()
    
And attributes:

    1. f.mode
    2. f.closed

````python
# Reading Files via 'with' context manager
# not only file, but also other resources, like threads

with open('test.txt', 'r') as f:
    
    print(f'File object:\t {f.name}')
    print(f'---> Open Mode:\t {f.mode}\n')

    # File contents in one string variable:
    file_contents = f.read()
    print(f'The whole file was read into a string variable at a time:\n{file_contents}\n')


# Context Manager closes the file automatically once out of the with-block
print(f'---> Context Manager exits. Is the file closed?\t {f.closed}\n')

try:
    # read file when closed
    no_contents = f.read()

except ValueError: 
    print('f.read(): I/O operation on closed file.\n')
except:
    print('Something else went wrong.\n')

    
words = file_contents.split(' ')
print(f'View words in the "test.txt" file:\n{words}\n')

word_count = len(words)
print(f'There are {word_count} words in the "test.txt" file.\n')

````

### 2.2 More Examples about File Reading Methods 

````python
# More reading methods
with open('test.txt', 'r') as f:
    
    # read file contents in a list of lines:
    f_contents = f.readlines()
    print(f'read lines into a list at a time:\n{f_contents}\n')

    # rewind to the head of the file
    f.seek(0)

    # read one line at a time:
    f_contents = f.readline()
    print(f'read first line into a list at a time:\n{f_contents}')

    f_contents = f.readline()
    print(f'read 2nd line into a list next time:\n{f_contents}')

    # mute "\n" effect in the buffer
    f_contents = f.readline()
    print(f'read 3rd line into a list, ignore \\n:\n{f_contents}', end='')
    f_contents = f.readline()
    print(f'read 4th line into a list, ignore \\n:\n{f_contents}', end='')

    # Iterating in a for-loop, line by line each loop till the end of file
    for line in f:
        print(f'Iterating in a for-loop, line by line each loop till the end...\n->{line}<-\n', end='')

    # rewind to the begining of file (BOF)
    f.seek(0)

    # Iterating in a while-loop in bytes till the end of file (EOF)
    size_to_read = 100
    f_contents = f.read(size_to_read)
    while len(f_contents) > 0:

        print(f'\nIterating in a while-loop, {size_to_read} bytes each loop till the end...\n->{f_contents}<-\n', end='')
        f_contents = f.read(size_to_read)

    # rewind to the head of the file
    f.seek(0)

    # read in small chunks, 10 characters per chunk:
    size_to_read = 10
    f_contents = f.read(size_to_read)
    print(f'read {size_to_read} bytes, ignore \\n:\n->{f_contents}<-\n', end='')
    print(f'Current position of the file objet = {f.tell()}\n')


    f.seek(0)
    f_contents = f.read(size_to_read)
    print(f'back to BOF, read {size_to_read} bytes, ignore \\n:\n->{f_contents}<-\n', end='')

    print(f'Current position of the file objet = {f.tell()}\n')
````

### 2.4 Copy a File using Context Manager

````python
# Copy a File to Another:
with open("test.txt", "r") as rf:
    with open("test_copy.txt", "w") as wf:
        for line in rf:
            wf.write(line)

file_read = "test_copy.txt"
with open(file_read, "r") as rf:
    print(f'opening {file_read} file:\n')
    for line in rf:
        print(f'Iterating line by line till the end...\n->{line}<-\n', end='')


# Copy an image: read/write in binary
with open("Olisan.JPG", "rb") as rf:
    with open("Olisan_copy.JPG", "wb") as wf:
        for line in rf:
            wf.write(line)

# Copy an image: in chunks
with open("Olisan.JPG", "rb") as rf:
    with open("Olisan_copy2.JPG", "wb") as wf:
        chunk_size = 4096
        rf_chunk = rf.read(chunk_size)
        
        while len(rf_chunk) > 0:
            wf.write(rf_chunk)
            rf_chunk = rf.read(chunk_size)

# Show all the images
from PIL import Image
img = Image.open('Olisan.JPG')
img.show()
img_copy = Image.open('Olisan_copy.JPG')
img_copy.show()
img_copy2 = Image.open('Olisan_copy2.JPG')
img_copy2.show()
````

## 3. Use try-except-else-finally for error handling

Use try/except to handle runtime errors gracefully.
It prevents your code from crashing unexpectedly.



````python
try:
    # Code that might raise an exception
    result = 10 / 0
except ZeroDivisionError:
    # Handle the specific exception
    print("Error: Cannot divide by zero")
else:
    # Code to run if no exception occurs
    print("else: Division successful")
finally:
    # Code to run regardless of whether an exception occurred or not
    print("finally: This will always execute")
````

## 4. Use enumerate for counting

When looping through lists, use enumerate() instead of manual counters.
It automatically gives you both index and value, improving clarity.



````python
# Old way using counter
names = ['Corey', 'Michael', 'Travis', 'Dave']

index = 0
for name in names:
    print(index, name)
    index += 1
````

````python
# cleaner code with enumerator
names = ['Corey', 'Michael', 'Travis', 'Dave']

for index, name in enumerate(names):
    print(index, name)
````

````python
# Another cleaner code with enumerator, starting index from 1
names = ['Corey', 'Michael', 'Travis', 'Dave']

# changing starting index from 1
for index, name in enumerate(names, start=1):
    print(index, name)
````

## 5. Use zip for multiple objects

Combine multiple lists cleanly with zip().
It’s perfect for looping through related data side by side.


````python
names = ['Corey', 'Michael', 'Travis', 'Dave']
heroes = ['Spiderman', 'Superman', 'Deadpool', 'Batman']
for index, name in enumerate(names):
    hero = heroes[index]
    print(f'{name} is actually {hero}')
````

````python
names = ['Corey', 'Michael', 'Travis', 'Dave']
heroes = ['Spiderman', 'Superman', 'Deadpool', 'Batman']

# cleaner code with zip
for name, hero in zip(names, heroes):
    print(f'{name} is actually {hero}')
````

````python
names = ['Corey', 'Michael', 'Travis', 'Dave']
heroes = ['Spiderman', 'Superman', 'Deadpool', 'Batman']
universes = ['Marvel', 'DC', 'Tokyo', 'Taiwan']

# cleaner code with zip more than 2 lists
for name, hero, universe in zip(names, heroes, universes):
    print(f'{name} is actually {hero} from {universe}')
````

## 6. Use unpacking to assign an object to variables


It allows you to assign values to variables in a single line of code.


````python
# Basic unpacking with tuple
a, b = (1, 2)
print(f"a: {a}")
print(f"b: {b}")
````

````python
# unpacking a list
x, y = [10, 20]
print(f"x: {x}, y: {y}")
````

````python
# Unpack in function arguments
def add(a, b, c):
    return a + b + c

args = (1, 2, 3)
print(f"add(*args) = {add(*args)}")

    
````

````python
# unpack a string which is an iterable, consisting a sequence of characters
a, b, c = "abc"
print(f"a={a}, b={b}, c={c}")
````

````python
# Unpacking a dictionary, using ** to unpack key-value pairs
def greet(name, age):
    print(f"Hello, {name}. You are {age} years old.")

person = {"name": "John", "age": 30}
greet(**person)

# same as
greet(name="John", age=30)
````

````python
# unpacking keys or valures from a dictionary
d = {'a': 1, 'b': 2, 'c': 3}

k1, k2, k3 = d.keys()
v1, v2, v3 = d.values()

print(f"k1={k1}, k2={k2}, k3={k3}")
print(f"v1={v1}, v2={v2}, v3={v3}")
````

````python
# Use * for extended unpacking (Python3+)
a, b, *c, d = (1, 2, 3, 4, 5, 6)
print(a)
print(b)
print(c)
print(d)
````

````python
a, b, *_, d = (1, 2, 3, 4, 5, 6)
print(a)
print(b)
print(d)
````

````python
# nested unpacking
coords =[(1,2),(3,4),(5,6)]
x,y = zip(*coords)
print(f"x: {x}")
print(f"y: {y}")
````

````python
# more complex nested list
data = ['Michael', (25, 'Engineer')]

# unpacking
name, (age, profession) = data
print(f"name: {name}, age: {age}, profession: {profession}")
````

## 7. Setattr/Getattr to dynamically set/get attributes of an object

Dynamically setting and getting attributes of an object.

````python
class Person():
    pass

person = Person()

# setting attributes and values on the fly
person.first = "Michael"
person.last = "Kao"

print(f'{person.first} {person.last}')
````

````python
class Person():
    pass

person = Person()

# setting an attribute with the value of a variable on the fly
key = "first"
value = "Michael"  
person.last = "Kao"

setattr(person, key, value)

print(f'{person.first} {person.last}')
````

````python
class Person():
    pass

person = Person()

# setting an attribute by the value of a variable on the fly
key = "first"
value = "Michael"  
person.last = "Kao"

setattr(person, key, value)

# getting an attribute value by the value of a variable
first = getattr(person, key)
print(first)
````

````python
class Person():
    pass

person = Person()

person_dict = {'first':'Michael', 'last':'Kao'}

for key, value in person_dict.items():
    setattr(person, key, value)

print(f'{person.first} {person.last}')

for key in person_dict.keys():
    print(getattr(person, key))
````

## 11. Iterables for iteration

    Iterable is an object, that one can iterate over. It generates an Iterator when passed to iter() method. 
    
    An iterator is an object, which is used to iterate over an iterable object using the __next__() method. 
    
    Iterators have the __next__() method, which returns the next item of the object. Note that every iterator is also an iterable, but not every iterable is an iterator. 
    
    For example, a list is iterable but a list is not an iterator. An iterator can be created from an iterable by using the function iter(). To make this possible, the class of an object needs either a method __iter__, which returns an iterator, or a __getitem__ method with sequential indexes starting with 0. 

````python
for element in ['c123', 142_857, 'Michael', 'Kao']:
    print(element, end=" ")
print()

for element in ('c123', 142_857, 'Michael', 'Kao'):
    print(element, end=" ")
print()

for letter in 'Michael Kao':
    print(letter, end=" ")
print()

for byte in b'Binary':
    print(byte, end=" ")
print()

# All objects, shown above, are iterables
# However, the integer type is not literable, as follows
c = 142_857

# must convert to a string to make  iteger iterable
for digit in str(c):
    print(digit, end=" ")
print()

# or convert to a list to make  iteger iterable
digits = [int(d) for d in str(c)]
for digit in digits:
    print(digit, end=" ")
print()

# make a deck of cards iterable?
import itertools
ranks = list(range(2, 11)) + ["J", "Q", "K", "A"]
ranks = [str(rank) for rank in ranks]

print(f'Ranks: {ranks}')

suits = ["Spades", "Hearts", "Dimonds", "Clubs"]
deck = [card for card in itertools.product(suits, ranks)]

for (index, card) in enumerate(deck):
    print(f'{1+index}: {card}')

hands = [hand for hand in itertools.combinations(deck, 5)]
print(f'The number of 5-card poker has {len(hands):,} combinations.')
````
